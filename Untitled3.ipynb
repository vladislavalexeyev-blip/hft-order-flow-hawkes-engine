```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
n_ticks = 50000

# 1. Генерация микросекундных L2-данных (Best Bid / Best Ask + Volumes)
time_stamps = pd.date_range(start='2026-01-01 09:30:00', periods=n_ticks, freq='1ms')
price_deltas = np.random.choice([-0.01, 0.00, 0.01], size=n_ticks, p=[0.15, 0.70, 0.15])
mid_price = 100.0 + np.cumsum(price_deltas)

spread = 0.01
bid_price = np.round(mid_price - spread / 2, 2)
ask_price = np.round(mid_price + spread / 2, 2)

bid_size = np.random.randint(100, 2000, size=n_ticks)
ask_size = np.random.randint(100, 2000, size=n_ticks)

df_l2 = pd.DataFrame({
    'timestamp': time_stamps,
    'bid_price': bid_price,
    'ask_price': ask_price,
    'bid_size': bid_size,
    'ask_size': ask_size
})

# 2. Векторизованный расчет Order Flow Imbalance (OFI)
def compute_ofi(df):
    dB_price = df['bid_price'].diff()
    dA_price = df['ask_price'].diff()

    bid_size_diff = df['bid_size'].diff()
    ask_size_diff = df['ask_size'].diff()

    # Delta Bid Volume
    delta_v_b = np.where(dB_price > 0, df['bid_size'],
                np.where(dB_price < 0, 0, bid_size_diff))

    # Delta Ask Volume
    delta_v_a = np.where(dA_price < 0, df['ask_size'],
                np.where(dA_price > 0, 0, ask_size_diff))

    return delta_v_b - delta_v_a

df_l2['OFI'] = compute_ofi(df_l2)
df_l2['Cumulative_OFI'] = df_l2['OFI'].cumsum()

# 3. Визуализация связи кумулятивного OFI и цены
fig, ax1 = plt.subplots(figsize=(12, 5))
ax2 = ax1.twinx()

ax1.plot(df_l2['timestamp'][:10000], df_l2['bid_price'][:10000], color='black', label='Bid Price', alpha=0.8)
ax2.plot(df_l2['timestamp'][:10000], df_l2['Cumulative_OFI'][:10000], color='blue', linestyle='--', label='Cumulative OFI', alpha=0.6)

ax1.set_xlabel('Timestamp')
ax1.set_ylabel('Price ($)', color='black')
ax2.set_ylabel('Cumulative OFI (Shares)', color='blue')
plt.title('L2 Market Microstructure: Mid-Price vs Order Flow Imbalance (OFI)')
plt.grid(True, alpha=0.2)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Параметры объема для бакетов
bucket_size = 20000        # Объем одного бакета V (20,000 акций)
n_buckets_window = 15      # Окно N бакетов для расчета роллинг-VPIN

# Классификация объемов: используем знак OFI для разделения на покупки и продажи
total_vol = df_l2['bid_size'] + df_l2['ask_size']
df_l2['total_volume'] = total_vol

df_l2['buy_volume'] = np.where(df_l2['OFI'] > 0, total_vol,
                       np.where(df_l2['OFI'] == 0, total_vol / 2, 0))
df_l2['sell_volume'] = total_vol - df_l2['buy_volume']

# 2. Нарезка данных наVolume Buckets
df_l2['cum_volume'] = df_l2['total_volume'].cumsum()
df_l2['bucket'] = (df_l2['cum_volume'] // bucket_size).astype(int)

bucket_stats = df_l2.groupby('bucket').agg(
    timestamp=('timestamp', 'last'),
    price=('bid_price', 'last'),
    buy_vol=('buy_volume', 'sum'),
    sell_vol=('sell_volume', 'sum')
).reset_index()

# 3. Расчет метрики токсичности VPIN
bucket_stats['order_imbalance'] = (bucket_stats['buy_vol'] - bucket_stats['sell_vol']).abs()
bucket_stats['VPIN'] = (
    bucket_stats['order_imbalance'].rolling(window=n_buckets_window).sum()
    / (n_buckets_window * bucket_size)
)

# 4. Визуализация индекса токсичности
fig, ax1 = plt.subplots(figsize=(12, 5))
ax2 = ax1.twinx()

ax1.plot(bucket_stats['timestamp'], bucket_stats['price'], color='black', label='Mid Price', alpha=0.8)
ax2.plot(bucket_stats['timestamp'], bucket_stats['VPIN'], color='red', label='VPIN Index', linewidth=1.5)
ax2.axhline(0.60, color='orange', linestyle='--', label='Toxicity Warning Threshold (0.60)')

ax1.set_xlabel('Timestamp')
ax1.set_ylabel('Price ($)', color='black')
ax2.set_ylabel('VPIN [0, 1]', color='red')
plt.title('Order Flow Toxicity Engine: VPIN vs Price Movement')
plt.grid(True, alpha=0.2)
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Фильтрация временных меток всплесков ликвидности (топ-5% крупных OFI движения)
threshold = df_l2['OFI'].abs().quantile(0.95)
shock_events = df_l2[df_l2['OFI'].abs() > threshold]
event_times = (shock_events['timestamp'] - df_l2['timestamp'].iloc[0]).dt.total_seconds().values

# 2. Параметры процесса Хоукса
mu = 2.0     # Базовая интенсивность
alpha = 1.2  # Сила самовозбуждения
beta = 1.5   # Скорость экспоненциального затухания
branching_ratio = alpha / beta

# 3. Расчет роллинг-интенсивности lambda(t)
t_grid = np.linspace(0, event_times[-1], 2000)
lambda_t = np.full_like(t_grid, mu, dtype=float)

for i, t in enumerate(t_grid):
    past_events = event_times[event_times < t]
    if len(past_events) > 0:
        lambda_t[i] += alpha * np.sum(np.exp(-beta * (t - past_events)))

# 4. Визуализация интенсивности каскадных атак
plt.figure(figsize=(12, 5))
plt.plot(t_grid, lambda_t, color='purple', linewidth=1.2, label=f'Hawkes Intensity $\lambda(t)$ ($\eta={branching_ratio:.2f}$)')
plt.axhline(mu / (1 - branching_ratio), color='darkred', linestyle='--', label=r'Critical Cluster Baseline $\frac{\mu}{1-\eta}$')
plt.scatter(event_times, np.full_like(event_times, mu), color='orange', s=12, zorder=5, alpha=0.6, label='Micro-shocks')

plt.title('Market Microstructure: Hawkes Self-Exciting Order Cascade')
plt.xlabel('Time (seconds from start)')
plt.ylabel('Event Intensity $\lambda(t)$ [Orders/sec]')
plt.grid(True, alpha=0.2)
plt.legend(loc='upper right')
plt.show()
```
